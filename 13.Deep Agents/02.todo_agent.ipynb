代办事项清单

```python
from typing import Annotated, Literal, NotRequired, TypedDict
from langchain.agents import AgentState


class Todo(TypedDict):
    """
    用于跟踪过复杂工作流进度的结构化任务项
    content:任务的简短和具体描述
    status:当前状态---待处理、进行中、已完成
    """
    content: str
    status: Literal["pending", "in_progress", "completed"]


def file_reducer(left, right):
    """
    将两个文件字典合并，右侧优化
    用作代理状态中文件字段的归约函数，允许对虚拟文件系统进行增量更新
    """
    if left is None:
        return right
    elif right is None:
        return left
    else:
        return {**left, **right}


class DeepAgentState(AgentState):
    """
    扩展的代理状态，包含任务跟踪和虚拟文件系统
    继承langgraph的AgentState并添加：
    todos:用于任务规划和进度跟踪的代办事项清单
    files:以字典形式存储的虚拟文件系统，将文件名映射到内容
    """
    todos: NotRequired[list[Todo]]
    files: Annotated[NotRequired[dict[str, str]], file_reducer]
```

```python
from langchain_core.messages import ToolMessage
from langchain_core.tools import InjectedToolCallId, tool
from langgraph.prebuilt import InjectedState
from langgraph.types import Command
from deep_agent_from_search.prompt import WRITE_TODOS_DESCRIPTION, TODO_USAGE_INSTRUCTIONS


@tool(description=WRITE_TODOS_DESCRIPTION)
def write_todos(
        todos: list[Todo],
        tool_call_id: Annotated[str, InjectedToolCallId]
) -> Command:
    """
    创建或更新代理的代办事项列表，用于任务规划和跟踪
    """
    return Command(
        update={
            "todos": todos,
            "messages": [
                ToolMessage(
                    f"更新任务清单:{todos}", tool_call_id=tool_call_id
                )
            ]
        }
    )


@tool(description="列出当前任务清单", parse_docstring=True)
def read_todos(
        state: Annotated[DeepAgentState, InjectedState]
) -> str:
    """
    从代理状态中读取当前的待办事项列表。

    Args:
        state: 包含当前 todo 列表的注入代理状态。

    Returns:
        当前 todo 列表的格式化字符串表示。
    """
    todos = state.get("todos", [])

    if not todos:
        return "当前没有任务清单"

    result = "任务列表：\n"

    for i, todo in enumerate(todos, 1):
        status_emoji = {
            "pending": "⏳",
            "in_progress": "🔄",
            "completed": "✅",
        }

        emoji = status_emoji.get(todo["status"], "?")

        result += (
            f"{i},{emoji}{todo['content']},({todo['status']})\n"
        )

    return result.strip()
```

```python
from IPython.display import Image, display
from langchain_openai.chat_models import ChatOpenAI
from langchain.agents import create_agent
from Config.load_key import open_key, base_url, base_model

search_result = "模型上下文协议（mcp）是有Anthropic开发的一项开放标准协议，旨在实现AI模型与外部系统（如工具，数据库及其他服务）之间的无缝链接。它充当一种标准化的通信层，是AI模型能够以一致且高效的方式访问并利用来自不同来源的数据，从根本上说，MCP通过提供一种统一的数据交换语言，简化了将AI助手与外部服务进行链接的过程"

SIMPLE_RESEARCH_INSTRUCTIONS = """
重要提示：只需调用一次web_search工具，并使用该工具提供的结果来回答用户的问题。
"""


@tool(description="搜索结果")
def web_search(
        query: str):
    return search_result


model = ChatOpenAI(
    model=base_model,
    api_key=open_key,
    base_url=base_url,
    temperature=0.0
)
agent = create_agent(
    model=model,
    tools=[read_todos, write_todos, web_search],
    system_prompt=TODO_USAGE_INSTRUCTIONS
                  + "\n\n"
                  + "=" * 80
                  + "\n\n" + SIMPLE_RESEARCH_INSTRUCTIONS,
    state_schema=DeepAgentState
)
display(Image(agent.get_graph(xray=True).draw_mermaid_png()))
```

```python
result = agent.stream({
    "messages": [
        {
            "role": "user",
            "content": "帮我总结一下MCP服务的含义"
        }
    ]
})
for chunk in result:
    print("\n==============================")

    for node_name, data in chunk.items():
        print(f"节点：{node_name}", end="", flush=True)
        if not data:
            continue
        messages = data.get("messages", [])
        for message in messages:
            # AI 消息
            if message.type == "ai":
                if message.tool_calls:
                    for tool_call in message.tool_calls:
                        print(
                            f"🔧 调用工具：{tool_call['name']}", end="", flush=True
                        )
                        print(
                            f"   参数：{tool_call['args']}", end="", flush=True
                        )
                elif message.content:
                    print(f"🤖 AI：{message.content}", end="", flush=True)
            # Tool 消息
            elif message.type == "tool":
                print(
                    f"🔨 工具结果：{message.name}", end="", flush=True
                )
                print(
                    f"   {message.content}", end="", flush=True
                )
```
